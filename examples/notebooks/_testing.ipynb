```python
import pandas as pd
import numpy as np
from pyMyriad import AnalysisTree
from pyMyriad.listing import gt_table
```

```python
# Create sample data and run analysis
np.random.seed(42)
df = pd.DataFrame(
    {
        "Gender": np.random.choice(["M", "F"], 200),
        "Country": np.random.choice(["US", "UK", "Canada"], 200),
        "Age": np.random.randint(18, 70, 200),
        "Income": np.random.normal(50000, 15000, 200),
    }
)

atree = (
    AnalysisTree()
    .split_by("df.Gender")
    .summarize_by(
        mean_income=lambda df: np.mean(df.Income),
        std_income=lambda df: np.std(df.Income)
    )
    .split_by("df.Country")
    .summarize_by(
        mean_income=lambda df: np.mean(df.Income),
        std_income=lambda df: np.std(df.Income),
    )
    .cross_analyze_by(mean_diff = "np.mean(df.Age) - np.mean(df.Income)")
)

result = atree.run(df)
```
